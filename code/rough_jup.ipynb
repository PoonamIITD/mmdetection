```python
from mmengine.runner import load_checkpoint
from mmdet.registry import MODELS
from mmcv.image import imread
from mmdet.structures import DetDataSample
# from mmdet.datasets.transforms import Compose
from mmengine.config import Config
import torch

# Paths to config and checkpoint
config_path = '../configs/mm_grounding_dino/grounding_dino_swin-t_finetune_8xb4_20e_cat.py'
checkpoint = '../grounding_dino_swin-t_pretrain_obj365_goldg_grit9m_v3det_20231204_095047-b448804b.pth'

# Load config and model
cfg = Config.fromfile(config_path)
model = MODELS.build(cfg.model)
load_checkpoint(model, checkpoint, map_location='cuda:2', strict=False)
model.to('cuda:2')
model.eval()

# Read image
img = imread('../data/cat/images/IMG_20211205_120756.jpg')

# Prepare data (you can also run full pipeline here)
# Debug forward pass
with torch.no_grad():
    result = model.forward([img], mode='predict')

```
